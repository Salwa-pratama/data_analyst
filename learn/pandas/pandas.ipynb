1. Import library

```python
import pandas as pd
import os
from dotenv import load_dotenv
```

```python
insurance_dataset = pd.read_csv("../../dataset/insurance.csv")

insurance_dataset.head()
```

```python
insurance_dataset.tail()
```

# export dataset ke jenis file lain

Export file csv ke csv

```python
# csv to csv
path = "../../dataset/"
file_name = "data_lain.csv"
insurance_dataset.to_csv(f"{path}{file_name}", index=False)

print(f"Dataset insurace.csv berhasil di export menjadi {file_name}")
print(f"Pada path : {path + file_name}")
```

Export csv ke excel

```python
file_name = "data_excel.xlsx"
insurance_dataset.to_excel(f"{path}{file_name}", index=False)

print(f"Dataset insurace.csv berhasil di export menjadi {file_name}")
print(f"Pada path : {path + file_name}")
```

Read Data HTML

```python
load_dotenv()
url = os.getenv("FDIC_URL")
data_html_fdic = pd.read_html(url)

df_bank = data_html_fdic[0]
```

Export data html ke data excel

```python
file_name = "data_bank.csv"
df_bank.to_csv(f"{path}{file_name}", index=False)
```

Export dari csv ke sql

```python
from sqlalchemy import create_engine
mesin = create_engine(f'sqlite:///{path}asuransi.db')
insurance_dataset.to_sql(f'tabel_asuransi', con=mesin, index=False, if_exists='replace')
print("Data asuransi berhasil di export menjadi sql")
```
